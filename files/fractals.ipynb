## The Koch Snowflake

The [**Koch Curve**](https://en.wikipedia.org/wiki/Koch_snowflake) first discovered by Swedish mathematician Helge von Koch is one of the first examples of a fractal. To draw a Koch curve we use the following steps:

1. Starting at stage 0 with a straight line of length $L$



2. Remove the middle third of the line and replace with two lines of length $\dfrac{L}{3}$ which form two sides of an equilateral triangle (so after we have now replaced the line with four lines)



3. Repeat steps 1. and 2. for each new line



The Python program in the code cell below defines a function called `koch_curve` which uses the following inputs:

- `x` and `y` are the $(x, y)$ co-ordinates of the end points of the lines
- `L` is the length of the initial line
- `angle` is the angle between the initial line and the horizontal
- `n` is the number of stages we are generating the Koch curve to

The function uses the value of $n$ to decide what to do. If $n = 0$ then we have finished with the current line segment so we calculate the co-ordinates of the endpoitn of the current line. If $n > 0$ then we replace the current line segments with four new lines. The first and last line use the same angle as the initial line, but the angle for the middle two line segements are increased or decreased by $60^\circ$ so that they form the two sides of the equilateral triangle.

#### Activity

1. Run the code cell generate the Koch curve up to stage 1 
2. Change the code to generate the Koch curve up to stage 2, 3 etc. 
3. What do you notice as of stages gets larger (e.g., $n > 8$)?

```python
import numpy as np
import matplotlib.pyplot as plt

def koch_curve(x, y, L, angle, n):
    if n == 0:
        # Add the end point of the line
        x.append(x[-1] + L * np.cos(np.radians(angle)))
        y.append(y[-1] + L * np.sin(np.radians(angle)))
    else:
        # Add lines to the curve
        koch_curve(x, y, L / 3, angle, n - 1)      # add line
        koch_curve(x, y, L / 3, angle + 60, n - 1) # turn 60 degrees anti-clockwise and add line
        koch_curve(x, y, L / 3, angle - 60, n - 1) # turn 60 degrees clockwise and add line
        koch_curve(x, y, L / 3, angle, n - 1)      # add line


# Define Koch curve inputs
x = [0]     # start point x co-ordinate
y = [0]     # start point y co-ordinate
L = 1       # length of the initial line
angle = 0   # angle between the initial line and the horizontal
n = 1       # number of stages generated

# Generate Koch curve
koch_curve(x, y, L, angle, n)

# Plot Koch curve
fig, ax = plt.subplots()
plt.plot(x, y, color="blue", linewidth=0.5)
ax.axis("off")
ax.set_aspect("equal")
```

### The Koch snowflake

We can join three Koch curves together to form a fractal known as the **Koch snowflake**.



#### Activity

Run the code cell below to produce Koch snowflake up to stage 1. Experiment with increasing the number of stages the Koch snowflake is generated to.

```python
# Define Koch snowflake inputs
x = [0]     # start point x co-ordinate
y = [0]     # start point y co-ordinate
L = 1       # length of the initial line
angle = 0   # angle between the initial line and the horizontal
n = 1       # number of stages generated

# Generate the Koch curves
koch_curve(x, y, L, angle, n)

angle = angle - 120   # rotate angle 120 degrees clockwise
koch_curve(x, y, L, angle, n)

angle = angle - 120   # rotate angle 120 degrees clockwise
koch_curve(x, y, L, angle, n)

# Plot the curves
fig, ax = plt.subplots()
plt.plot(x, y, color="blue", linewidth=0.5)
plt.axis("equal")
plt.axis("off")
plt.show()
```

## The Mandelbrot Set

Perhaps the most famous and beautiful fractal is the [**Mandelbrot set**](https://en.wikipedia.org/wiki/Mandelbrot_set). It is generated by repeatedly applying the following simple formula to complex numbers:

$$ z_{n+1} = z_{n}^2 + c, $$

where $c = x + y\,i$ is a complex number, $i^2 = -1$ is the imaginary number and the starting value of $z_0$ is $0 + 0i$. We use this formula to calculate $z_1$ using $z_0$, and then again to calculate $z_2$ using $z_1$, repeating this many times (this is known as **iteration**)

\begin{align*}
    z_1 &= z_0^2 + c, \\
    z_2 &= z_1^2 + c, \\
    z_3 &= z_2^2 + c, \\
    &\vdots \\
\end{align*}

If the values of $z_n$ stay bounded (in practice this means if $|z_n| \leq 2$ for all iterations), then $c$ is said to be a member of the Mandelbrot set. If $|z_n|$ ever becomes larger than 2, then $c$ is not a member of the Mandelbrot set. The key idea is to see whether the values *"escape”* to infinity or stay small. Points inside the Mandelbrot set stay bounded, while points outside quickly grow very large.

For example, let's see if $c = 1 + 0i$ is in the Mandelbrot set

\begin{align*}
  z_1 &= z_0^2 + c = (0 + 0i)^2 + (1 + 0i) = 1 + 0i, & |z_1| &= \sqrt{1^2 + 0^2} = 1, \\
  z_2 &= z_1^2 + c = (1 + 0i)^2 + (1 + 0i) = 2 + 0i, & |z_2| &= \sqrt{2^2 + 0^2} = 2, \\
  z_3 &= z_2^2 + c = (2 + 0i)^2 + (1 + 0i) = 5 + 0i, & |z_3| &= \sqrt{5^2 + 0^2} = 5.
\end{align*}

Here $|z_3| > 2$ so $c = 1 + 0i$ is **not** in the Mandelbrot set. 

What about $c = 0 + i$?

\begin{align*}
  z_1 &= z_0^2 + c = (0 + 0i)^2 + (0 + i) = 0 + i, & |z_1| &= \sqrt{0^2 + 1^2} = 1, \\
  z_2 &= z_1^2 + c = (0 + i)^2 + (0 + i) = -1 + i, & |z_2| &= \sqrt{(-1)^2 + 1^2} = 1.414\ldots, \\
  z_3 &= z_2^2 + c = (-1 + i)^2 + (0 + i) = 0 - i, & |z_3| &= \sqrt{0^2 + (-1)^2} = 1, \\
  z_4 &= z_3^2 + c = (0 - i)^2 + (0 + i) = -1 + i, & |z_4| &= \sqrt{(-1)^2 + 1^2} = 1.414\ldots, \\
  &\vdots
\end{align*}

Here we see that the values start to repeat: $z_2 = z_4 = z_6 = \ldots$. This means the sequence gets stuck in a loop and will never grow larger. Since $|z_n| \leq 2$ for all iterations, $c = 0 + i$ **is** in the Mandelbrot set. We can draw the Mandelbrot set by plotting a point at coordinates $(x, y)$ for each value $c = x + y \, i$ that is a Mandelbrot set. 

The code cell below performs these iterations for many values of $c$. Points in the Mandelbrot set are shown as black, while points outside the set are coloured according to how quickly they “escape” (this is called the escape time).

#### Activity 

Run the code cell below to generate the Mandelbrot set.

```python
def mandelbrot(x_min, x_max, y_min, y_max, max_iter):

  # Setup arrays
  X, Y = np.meshgrid(np.linspace(x_min, x_max, 800), np.linspace(y_max, y_min, 800))
  C = X + Y * 1j                          # values of c = x + yi
  Z = np.zeros_like(C)                    # values of z = 0 + 0i
  escape = np.zeros(C.shape, dtype=int)   # the escape number

  # Perform iterations
  for n in range(max_iter):
    mask = np.abs(Z) <= 2            # find the points still inside Mandelbrot set
    Z[mask] = Z[mask]**2 + C[mask]   # calculate z_n+1 = z_n^2 + c
    escape[mask] = n                 # update the escape number

  # set escape number for points inside the Mandelbrot to -1 for plotting
  mask = np.abs(Z) <= 2
  escape[mask] = -1           

  return escape


# Define Mandelbrot set parameters
x_min = -2     # min value of x
x_max = 1      # max value of x
y_min = -1.5   # min value of y
y_max = 1.5    # max value of y
max_iter = 30  # maximum number of iterations computed

# Generate Mandelbrot set
mandelbrot_set = mandelbrot(x_min, x_max, y_min, y_max, max_iter)

# Plot Mandelbrot set
cmap = plt.cm.jet.copy()
cmap.set_under('black')
plt.figure(figsize=(8, 8))
plt.imshow(mandelbrot_set, cmap=cmap, vmin=0, vmax=max_iter, extent=(x_min, x_max, y_min, y_max))
plt.xlabel('$x$')
plt.ylabel('$y$')
plt.show()
```
